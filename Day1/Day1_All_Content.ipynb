# 2. Perceptrons & Neurons

### Demystifying Perceptrons and Neurons
#### Perceptrons:
   - Foundational building block of neural networks and deep learning models.
   - It acts as a binary linear classifier, determining if an input belongs to one class or another.
   - Performs binary classification by computing a weighted sum of inputs and applying an activation function.
   - Introduced by Frank Rosenblatt in 1958.


#### Logic Gates Interpretation: 
  - Perceptrons can be used to model basic logic gates such as AND, OR, and NOT.
  - For instance, with appropriately adjusted weights, a perceptron can emulate the AND gate. Iftwo input values are '1' (true), the perceptron outputs '1' and '0' otherwise.
  - By adjusting the weights and bias, a perceptron can mimic the behavior of these gates.
  - This intrinsic capacity of perceptrons to reproduce logical operations underscores their poweras fundamental computational units in neural networks.

#### Linear Decision Boundaries:
- The perceptron makes its decision based on a linear function of the input. Essentially, if the weighted sum of the input surpasses a certain threshold, the perceptron activates; otherwise, it remains inactive. 
       
- In a 2D input space, this decision mechanism is represented as a line. For higher-dimensional inputs, it generalises to a hyperplane. 
- This line or hyperplane is termed the "decision boundary," delineating the regions corresponding to different output classes.


- Perceptrons are limited to linearly separable problems. If the data cannot be separated by a straight line (or hyperplane in higher dimensions), a single-layer perceptron will fail to classify it correctly. This limitation led to the development of multi-layer perceptrons (MLPs) which can handle non-linear decision boundaries through multiple layers and non-linear activation functions.

#### Perceptrons Layers:
- input layer: receives the input features.
- hidden layer(s): perform computations and extract features (can have multiple layers).
- output layer: produces the final prediction or output.

#### What is Perceptrons?
A **perceptron** is a type of artificial neuron used in machine learning for **binary classification** tasks.

#### Neurons:
  - Analogous to the perceptron, neurons compute a weighted sum of their inputs and pass the result through an activation function.
  - A neuron is the fundamental unit of a neural network, inspired by biological neurons.
  - It receives input signals, processes them using a weighted sum and an activation function, and produces an output signal.
  - Neurons can be organized into layers to form complex neural networks.

#### Perceptrons vs Neurons:
  - Perceptrons are a simplified model of neurons, primarily used for binary classification.
  - Neurons in modern neural networks are more flexible, using various activation functions and capable of handling complex, non-linear relationships.
  - While a single perceptron can only solve linearly separable problems, networks of neurons (multi-layer perceptrons) can approximate any continuous function given sufficient neurons and layers.

#### What are Connections and weights?
- Connections are the links between neurons in different layers of a neural network.
- Weights are numerical values assigned to these connections, determining the strength, Direction, and influence of the input signal on the neuron's output.
- Connections and weights are fundamental components of neural networks, enabling the network to learn and adapt by adjusting the weights during training.
- During Training, Weights are adjusted by using optimization techniques such as gradient descent to minimize the error between the predicted and actual outputs.
- This process allows the neural network to learn from data and improve its performance over time.

#### Heartbeat of Neurons:
#### Activation function:
  - An activation function is a mathematical function applied to the output of a neuron.
    - It introduces non-linearity into the network, allowing it to learn complex patterns.
    - Common activation functions include Sigmoid, Tanh, and ReLU (Rectified Linear Unit).
    - without these activation functions, neural networks would behave like linear models, limiting their ability to capture complex relationships in the data.
    - Non-Linearity :Allows neural networks to model complex relationships between inputs and outputs.
    - Thresholding:
        - Thresholding is a process where the output of a neuron is compared to a certain threshold value.
        - If the output exceeds the threshold, the neuron is activated; otherwise, it remains inactive.
        - This mechanism helps in controlling the firing of neurons and can be used to introduce non-linearity in the network.
    - Differentiability:
        - As deep learning models rely on gradient-based optimization techniques like gradient descent, the activation functions used should be differentiable.
        - Differentiability ensures that gradients can be computed for backpropagation, enabling the network to learn effectively.


#### Common Activation Functions:

- Sigmoid:
    - Maps input values to a range between 0 and 1.
    - Often used in the output layer for binary classification problems.
    - Suffers from the vanishing gradient problem, especially in deep networks.  This is because for very high or very low values of x, the gradient is almost zero.
    Equation: σ(x) = 1 / (1 + e^(-x))

- Tanh (Hyperbolic Tangent):
    - Maps input values to a range between -1 and 1.
    - Zero-centered, which can help with optimization.
    - Suffers from the vanishing gradient problem similar to Sigmoid.
    Equation: tanh(x) = (e^x - e^(-x)) / (e^x + e^(-x))

- ReLU (Rectified Linear Unit):
    - Outputs the input directly if it is positive; otherwise, it outputs zero.
    - Widely used due to its simplicity and effectiveness in deep networks.
    - Helps mitigate the vanishing gradient problem, allowing for better training of deep networks.
    Equation: ReLU(x) = max(0, x)

- Leaky ReLU:
    - Similar to ReLU, but allows a small, non-zero gradient when the input is negative.
    - Helps address the "dying ReLU" problem where neurons can become inactive and stop learning.
    Equation: LeakyReLU(x) = max(0.01x, x)

### what is a bias?
- A bias is an additional parameter in a neuron that allows the activation function to be shifted to the left or right.
- It helps the model to fit the data better by providing more flexibility.
- Mathematically, for a neuron with input x and weight w, the output is calculated as:
  z = w * x + b
  where b is the bias term.


### what is backpropagation?

- Backpropagation is a supervised learning algorithm used for training artificial neural networks.
- It calculates the gradient of the loss function with respect to each weight by the chain rule, allowing the network to update the weights to minimize the loss.
- The process involves two main steps:
  1. Forward pass: Compute the output of the network and the loss.
  2. Backward pass: Propagate the error backward through the network to compute gradients and update weights.

# 4. PyTorch Basics (`O01_pytorch.py`)


````python
'''
nvidia-smi : Check the status of NVIDIA GPU and its memory usage.

pytorch installation steps:
1. Check NVIDIA GPU status: nvidia-smi
2. Install PyTorch: pip install torch torchvision torchaudio
3. Verify installation: python -c "import torch; print(torch.__version__)"

4. python -c "import struct; print(struct.calcsize('P') * 8)" : Check the system architecture (32-bit or 64-bit).
'''

'''
what is PyTorch? 
- PyTorch is an open-source deep learning framework developed by Facebook's AI Research lab. 
- It provides a flexible and efficient platform for building and training neural networks, supporting both CPU and GPU computations.
- It is widely used in both academia and industry for research and production purposes due to its dynamic computation graph and ease of use.
'''

'''
what are tensors in PyTorch?
- Tensors are the fundamental data structures in PyTorch, similar to arrays or matrices in other numerical computing libraries.
- They can be used to represent inputs, outputs, and parameters of neural networks.
- PyTorch provides a wide range of tensor operations, including mathematical, logical, and indexing operations, which can be performed on both CPU and GPU.
- Tensors can also be easily converted between CPU and GPU, enabling efficient computation on different hardware.
'''

'''
Hello, this is a basic PyTorch script.
Each model has 1 input, and learns to extract formula for converting celsius to fahrenheit.
z = w.x + b
# where w is the weight, x is the input, and b is the bias.
F = 1.8 * C + 32
'''

# --- 1. Imports ---
# torch: Core library providing multi-dimensional tensors and automatic differentiation (autograd).
# torch.nn: Neural network module providing building blocks like layers (nn.Linear) and loss functions (nn.MSELoss).
import torch
import torch.nn as nn

# --- 2. Device Configuration ---
# Select NVIDIA CUDA GPU if available; otherwise fall back to CPU.
# Both the model and all data tensors must reside on the same device to avoid runtime errors.
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Print the installed PyTorch version
print(torch.__version__)

# --- 3. Training Data (Tensors) ---
# Each scalar value is wrapped in [...] to create a 2D tensor of shape (9, 1) = (num_samples, num_features),
# which is required by nn.Linear (expects a batch dimension first, followed by feature columns).
# dtype=torch.float32 uses standard 32-bit floating-point precision for weights and gradients.
# device=device places the tensors directly onto the selected CPU or GPU.
celsius = torch.tensor(
    [[-20.0], [-10.0], [0.0], [10.0], [20.0], [25.0], [30.0], [37.0], [100.0]],
    dtype=torch.float32,
    device=device
)

# Corresponding target Fahrenheit values, also with shape (9, 1) and dtype float32
fahrenheit = torch.tensor(
    [[-4.0], [14.0], [32.0], [50.0], [68.0], [77.0], [86.0], [98.6], [212.0]],
    dtype=torch.float32,
    device=device
)

# --- 4. Model Definition ---
# Single linear neuron: computes y = w * x + b (F = w * C + b).
# Initializes 1 learnable weight (w) and 1 learnable bias (b) with random values, then moves them to device.
model = nn.Linear(in_features=1, out_features=1)
model.to(device)

# --- 5. Loss Function & Optimizer ---
# MSELoss (Mean Squared Error) measures the average squared difference between predictions and targets.
# It is smooth/differentiable and penalizes larger errors more heavily.
loss_fn = nn.MSELoss()

# SGD (Stochastic Gradient Descent) updates each parameter in model.parameters() via: param = param - lr * grad.
# lr=0.0001 is kept small because input values reach 100.0, making MSE gradients large;
# a larger learning rate would overshoot the minimum and diverge (producing NaN).
optimizer = torch.optim.SGD(model.parameters(), lr=0.0001)

# --- 6. Training Loop (5 Core PyTorch Steps) ---
# Each epoch is one full pass over all 9 samples; many iterations are needed because
# the small learning rate causes the bias (b) to converge gradually toward 32.0.
for epoch in range(1, 50000):
    # Step 1: Clear gradients from the previous iteration (PyTorch accumulates gradients by default)
    optimizer.zero_grad()

    # Step 2: Forward pass — compute predicted Fahrenheit values by passing inputs through the model
    predictions = model(celsius)

    # Step 3: Compute scalar MSE loss comparing predictions to actual Fahrenheit targets
    loss = loss_fn(predictions, fahrenheit)

    # Step 4: Backpropagation — apply the chain rule via autograd to compute d(loss)/dw and d(loss)/db
    loss.backward()

    # Step 5: Update model weight and bias in the opposite direction of their gradients
    optimizer.step()

    # Log progress every 1,000 epochs; .item() extracts a plain Python float from the 0D loss tensor
    if epoch % 1000 == 0:
        print(f"Epoch {epoch:4d} | Loss: {loss.item():.6f}")

# --- 7. Inspect Learned Parameters ---
# .item() converts single-element tensors into Python floats.
# After training, learned_w should approach 1.8000 and learned_b should approach 32.0000.
learned_w = model.weight.item()
learned_b = model.bias.item()
print(f"\nLearned equation: F = {learned_w:.4f} * C + {learned_b:.4f}")

# --- 8. Evaluation / Inference ---
# Switch the model to evaluation mode (good practice; disables Dropout/BatchNorm training behavior in larger models)
model.eval()

# torch.no_grad() disables gradient tracking during inference, saving memory and speeding up computation.
with torch.no_grad():
    # Shape (1, 1) = (batch_size=1, in_features=1); 40°C should predict ~104.00°F
    test_celsius = torch.tensor([[40.0]], dtype=torch.float32, device=device)
    test_fahrenheit = model(test_celsius)
    print(f"Predicted Fahrenheit for 40°C: {test_fahrenheit.item():.2f}")
````

# 5. PyTorch Assignment (`O02_pytorch_assignment.py`)


````python

'''
Hello, this is a basic PyTorch script.
Each model has 1 input, and learns to extract formula for converting celsius to fahrenheit.
z = w.x + b
# where w is the weight, x is the input, and b is the bias.
F = 3 * C + 100
'''

import torch
import torch.nn as nn

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# print torch version
print(torch.__version__)

# create a celsius tensor
celsius = torch.tensor(
    [[-20.0], [-10.0], [0.0], [10.0], [20.0], [25.0], [30.0], [37.0], [100.0]], 
                dtype=torch.float32, 
                device=device
                )

fahrenheit = torch.tensor(
    [[40.0], [70.0], [100.0], [130.0], [160.0], [175.0], [190.0], [211.0], [400.0]],
    dtype=torch.float32,
    device=device
)

# 1 input feature -> 1 output value (learns weight w and bias b)
model = nn.Linear(in_features=1, out_features=1)
model.to(device) 

loss_fn = nn.MSELoss() # Mean Squared Error loss for regression tasks
optimizer = torch.optim.SGD(model.parameters(), lr=0.0001) # Stochastic Gradient Descent optimizer

# Train the model
for epoch in range(1, 50000):  # number of training iterations
    optimizer.zero_grad() # 1. Clearing previous gradients
    # Forward pass: compute predicted y by passing x to the model
    predictions = model(celsius) # 2. forward pass

    # Compute and print loss
    loss = loss_fn(predictions, fahrenheit) # 3. compute loss (MSE)

    loss.backward() # 4. backpropogation (compute gradients)
    optimizer.step() # 5. update weights and bias

    if epoch % 1000 == 0:
        print(f"Epoch {epoch:4d} | Loss: {loss.item():.6f}")

# Inspect learned parameters
learned_w = model.weight.item()
learned_b = model.bias.item()
print(f"\nLearned equation: F = {learned_w:.4f} * C + {learned_b:.4f}")

# Test on an unseen value (40°C should be 220.0°F)
model.eval()
# torch.no_grad() context is used to disable gradient calculation for inference, which reduces memory usage and speeds up computations.
with torch.no_grad():
    test_celsius = torch.tensor([[40.0]], dtype=torch.float32, device=device)
    test_fahrenheit = model(test_celsius)
    print(f"Predicted Fahrenheit for 40°C: {test_fahrenheit.item():.2f}")
````

````python
'''
what is TensorFlow? 
- TensorFlow is an open-source deep learning framework developed by Google.
- It provides a flexible and efficient platform for building and training neural networks, supporting both CPU and GPU computations.
- It is widely used in both academia and industry for research and production purposes due to its computational graph and ease of use.
'''
'''
What is keras?
- Keras is an open-source deep learning API written in Python, running on top of TensorFlow.
- It provides a high-level interface for building and training neural networks, making it easier to experiment with deep learning models.
- Keras is known for its user-friendly and modular design, allowing for quick prototyping and experimentation.
'''
'''
Hello, this is a basic TensorFlow script.
Each model has 1 input, and learns to extract formula for converting celsius to fahrenheit.
z = w.x + b
# where w is the weight, x is the input, and b is the bias.
F = 1.8 * C + 32
'''

'''
installation:
- You can install TensorFlow using pip:
  ```
  pip install tensorflow
  ```
'''

'''
In PowerShell (ctrl + `), you can set the environment variables as follows:
$env:TF_CPP_MIN_LOG_LEVEL="2"
$env:TF_ENABLE_ONEDNN_OPTS="0"
'''
import os
os.environ.setdefault("TF_CPP_MIN_LOG_LEVEL", "2") # Suppress TensorFlow logging messages (INFO and WARNING) but show errors
os.environ.setdefault("TF_ENABLE_ONEDNN_OPTS", "0") # Disable oneDNN optimizations for TensorFlow

import numpy as np
import tensorflow as tf



# $env:F_ENABLE_ONEDNN_OPTS=0 # Disable oneDNN optimizations for TensorFlow
# print TensorFlow version
print(tf.__version__)

# Check if GPU is available
print("TensorFlow version:", tf.__version__)
gpus = tf.config.list_physical_devices("GPU")
print("GPUs available:", len(gpus), "| Running on CPU" if not gpus else "| Running on GPU")

# Explicit Celsius (input) and Fahrenheit (target) pairs: F = 1.8 * C + 32
celsius = np.array(
        [-20.0, -10.0, 0.0, 10.0, 20.0, 25.0, 30.0, 37.0, 100.0],
        dtype=np.float32,
    ).reshape(-1, 1)

fahrenheit = np.array(
        [-4.0, 14.0, 32.0, 50.0, 68.0, 77.0, 86.0, 98.6, 212.0],
        dtype=np.float32,
    ).reshape(-1, 1)

# Single Dense neuron with 1 input and 1 output (linear activation by default)
model = tf.keras.Sequential([ # Sequential model with a single input and output layer
        tf.keras.Input(shape=(1,)),# Input layer with 1 feature (Celsius)
        tf.keras.layers.Dense(units=1),# Single neuron with 1 input and 1 output
    ])

# Compile the model with Stochastic Gradient Descent optimizer and Mean Squared Error loss function
model.compile(
        optimizer=tf.keras.optimizers.SGD(learning_rate=0.0001),
        loss="mean_squared_error",
    )

# Train in stages so you can see loss decreasing every 1,000 epochs
total_epochs = 5000
step_epochs = 1000
for start_epoch in range(0, total_epochs, step_epochs):
        history = model.fit(
            celsius,
            fahrenheit,
            epochs=step_epochs,
            verbose=0,
        )
        end_epoch = start_epoch + step_epochs
        final_loss = history.history["loss"][-1]
        print(f"Epoch {end_epoch:4d}/{total_epochs} | MSE Loss: {final_loss:.6f}")

# Inspect learned parameters
weights, bias = model.layers[0].get_weights()
learned_w = float(weights[0][0])
learned_b = float(bias[0])
print(f"\nLearned equation: F = {learned_w:.4f} * C + {learned_b:.4f}")
print("Target equation:  F = 1.8000 * C + 32.0000")

# Test on unseen Celsius values
test_temps = np.array([[15.0], [40.0], [75.0]], dtype=np.float32)
predictions = model.predict(test_temps, verbose=0).flatten()
print("\nTest Predictions:")
for c_val, pred_f in zip(test_temps.flatten(), predictions):
        expected_f = 1.8 * float(c_val) + 32.0
        print(f"  {c_val:5.1f}°C -> Predicted: {pred_f:6.2f}°F | Expected: {expected_f:6.2f}°F")
````
